Assignment: Create a stack plot showing my weekly schedule.

```python
import matplotlib.pyplot as plt;
```

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'];
study = [3, 4, 3, 4, 3, 5, 4];
sleep = [8, 8, 7, 8, 8, 9, 9];
free_time = [2, 2, 3, 2, 3, 4, 4];
```

```python
plt.figure(figsize = (10, 6));

plt.stackplot(
    days,
    study,
    sleep,
    free_time,
    labels = ['Study', 'Sleep', 'Free Time']
);

plt.title('My Weekly Schedule');
plt.xlabel('Days');
plt.ylabel('Hours');
plt.legend(loc = 'upper left');

plt.show();
```
