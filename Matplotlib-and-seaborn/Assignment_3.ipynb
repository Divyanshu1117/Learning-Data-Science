Assignment:
Create a pie chart showing the market share of mobile OS (Android, iOS, others). Then recreate the same using a horizontal bar chart and observe which is easier to understand.

```python
import matplotlib.pyplot as plt;
```

```python
os = ['Android', 'iOS', 'Others'];
market_share = [70, 28, 2];
```

```python
# Pie Chart:-
plt.figure(figsize = (7, 5));

plt.pie(
    market_share,
    labels = os,
    autopct = '%1.1f%%',
    startangle = 90
);

plt.title('Mobile OS Market Share');
plt.show();
```

```python
# Horizontal Bar Chart:-
plt.figure(figsize = (7, 5));

plt.barh(os, market_share);

plt.xlabel('Market Share (%)');
plt.ylabel('Mobile OS');
plt.title('Mobile OS Market Share - Horizontal Bar Chart');

plt.show();
```
