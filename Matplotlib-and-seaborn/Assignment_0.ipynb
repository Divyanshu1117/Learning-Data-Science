Create a plot comparing Kohli, Rohit Sharma, and Sehwag across 10 years of hypothetical runs.
Use:
Labels
Legends
Colors
Line styles
One custom style

```python
import matplotlib.pyplot as plt;
```

```python
years = list(range(2015, 2025));
```

```python
virat_kohli = [1380, 1520, 1460, 1320, 1375, 1210, 1280, 1420, 1560, 1490];
rohit_sharma = [1250, 1390, 1310, 1480, 1360, 1180, 1290, 1510, 1430, 1540];
virender_sehwag = [1120, 1240, 1190, 1080, 1150, 1010, 980, 1050, 1100, 1070];
```

```python
plt.plot(years, virat_kohli, label = "Virat Kohli", color = "orange", linestyle = "--");
plt.plot(years, rohit_sharma, label = "Rohit Sharma", color = "green", linestyle = "-");
plt.plot(years, virender_sehwag, label = "Virender Sehwag", color = "black", linestyle = "--");
plt.xlabel("Year");
plt.ylabel("Runs");
plt.title("Hypothetical ODI Runs - 10 Years");
plt.legend();
plt.grid(True, linestyle = ":", alpha = 0.5);
plt.show();
```
