```python
plt.hist(ages, bins = bins, edgecolor = 'black');
plt.axvline(35, color = 'red', linestyle = '--', linewidth = 2, label = 'Average Age');
plt.title('Age Distribution of You Tube Viewers');
plt.xlabel('Age Group');
plt.ylabel('Number of Viewers');
plt.grid(True, linestyle = '--', alpha = 0.5);
plt.show();
```
