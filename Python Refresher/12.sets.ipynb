```python
# Set Operations.
set1 = {1, 2, 3, 4};
set2 = {3, 4, 5, 6};
```

```python
# Union.
print(set1 | set2);
```

```python
# Intersetion.
print(set1 & set2);
```

```python
# Difference.
print(set1 - set2);
```

```python
# Symmetric Difference.
print(set1 ^ set2);
```

```python
# Issubset.
sub = {1, 3};
```

```python
sub.issubset(set1)
```

```python
# Issuperset.
set1.issubset(sub)
```
