```python
# Python Zip:-
l1 = [1, 2, 4];
l2 = [6, 7, 8];
list(zip(l1, l2))
```

```python
# Using Python Lists:-
import numpy as np;
import time;

size = 1_000_000;

l1 = list(range(size));
l2 = list(range(size));

start = time.time();
add = [x + y for x,y in zip(l1,l2)];
end = time.time();
print(add[0:10]);

print(end - start);
```

```python
# Using Numpy Arrays:-
import numpy as np;
import time;

size = 1_000_000;

l1 = np.array(range(size));
l2 = np.array(range(size));

start = time.time();
add = l1 + l2;
end = time.time();
print(add[0:10]);

print(end - start);
```
