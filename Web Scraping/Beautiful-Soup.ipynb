```python
from bs4 import BeautifulSoup;
```

```python
with open("D:\Learning Data Science\Web Scraping\HTML'S\page1.html") as f:
    content = f.read();
```

```python
soup = BeautifulSoup(content, "html.parser");
```

```python
h3s = soup.find_all("h3");
```

```python
h3s
```

```python
for h3 in h3s:
    print(h3.find("a")['title']);
```

```python
articles = soup.select("article.product_pod");
```

```python
for article in articles:
    title = article.find("h3").find("a")["title"];
    print(title);
```

```python
for article in articles:
    title = article.find("h3").find("a")["title"];
    price = article.select_one("p.price_color").text.split("£")[1];
    print(price);
```

```python
items = [];
for article in articles:
    title = article.find("h3").find("a")["title"];
    price = article.select_one("p.price_color").text.split("£")[1];
    items.append([title, price]);
```

```python
items
```

```python
import pandas as pd;
```

```python
df = pd.DataFrame(items, columns = ["Book Title", "Price"]);
```

```python
print(df);
```

```python
items = [];
for article in articles:
    title = article.find("h3").find("a")["title"];
    price = article.select_one("p.price_color").text.split("£")[1];
    rating_element = article.select_one("p.star-rating");
    rating = rating_element['class'][1];
    items.append([title, price, rating]);
    print(rating);
```

```python
df = pd.DataFrame(items, columns = ["Book Title", "Price", "Rating"]);
```
